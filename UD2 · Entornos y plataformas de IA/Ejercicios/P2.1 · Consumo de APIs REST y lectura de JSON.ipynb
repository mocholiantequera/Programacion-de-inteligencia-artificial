```python
import requests
import time
import xml.etree.ElementTree as ET

URL = "https://datos.gob.es/apidata/catalog/dataset"
HEADERS = {"Accept": "application/json"}
TIMEOUT = 15
```

#### 1. Haz una petición GET y antes de mirar el contenido, imprime y comenta: - el código de estado, - la cabecera Content-Type de la respuesta, - el tamaño en bytes, - y el tiempo que ha tardado (respuesta.elapsed).

```python
respuesta = requests.get(URL, headers=HEADERS, timeout=TIMEOUT)
print ("Estado de la respuesta: ", respuesta.status_code) #Para ver el codigo de estado de la respuesta

print ("Encabezados de la respuesta: ", respuesta.headers) #Para ver los encabezados de la respuesta

print ("Ver el tamaño de la respuesta", len(respuesta.content)) #Para ver el tamaño de la respuesta

print ("Ver el tiempo de la respuesta", respuesta.elapsed.total_seconds()) #Para ver el tiempo de la respuesta
```

#### 2. Explica en una celda de texto qué te dice cada una de esas cuatro cosas y qué harías distinto si el Content-Type fuera application/xml en lugar de application/json.

1. Status_code: Nos da un número de tres cifras que dice si la petición fue bien(200) o falló. Los digitos que empiezan por 4 es culpa mía y los digitos que empiezan por 5xx/429 es temporal y se reintenta
   
2. Headers: Es un diccionarios con los metadatos de la respuesta. Lo clave es el Content-Type, que indica en qué formato viene el cuerpo para elegir el parser correcto.

3. len(respuesta.content): Bytes crudos del cuerpo, sin decodificar. Sirve para verificar que no está vacio y hacerse un idea del volumen a manejar

4. repuesta.elapsed.total_seconds(): Tiempo total de la petición en segundos. Sirve para fijar un timeout adecuado y detectar si el servidor está lento. 

Si fuera application/xml haria lo siguiente: la peticion seria igual, pero no podría usar respuesta.json(): parsearñia con xml.etree.ElementTree, navegando con etiquetas en vez de claves, y añadiría Accept: application/xml

#### 3. Repite la petición con un parámetro de paginación distinto y comprueba que el resultado cambia. Explica dónde van los parámetros de una petición GET y por qué no van en el cuerpo.

```python
URL2 = "https://datos.gob.es/apidata/catalog/dataset?page=2"
HEADERS = {"Accept": "application/json"}
TIMEOUT = 15


respuesta = requests.get(URL2, headers=HEADERS, timeout=TIMEOUT)
print ("Estado de la respuesta: ", respuesta.status_code) #Para ver el codigo de estado de la respuesta

print ("Encabezados de la respuesta: ", respuesta.headers) #Para ver los encabezados de la respuesta

print ("Ver el tamaño de la respuesta", len(respuesta.content)) #Para ver el tamaño de la respuesta

print ("Ver el tiempo de la respuesta", respuesta.elapsed.total_seconds()) #Para ver el tiempo de la respuesta

#El resultado de la peticion con esta paginacion cambia con respecto a la primera peticion, ya que en esta segunda peticion se esta solicitando la pagina 2 de los resultados.
```
