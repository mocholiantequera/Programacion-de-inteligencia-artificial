```python
# 1. Cárgalo con json.load() y describe su estructura: qué es objeto, qué es array y a qué profundidad está cada dato.
import json
from datetime import datetime

with open("ficheros/licitaciones_api.json", encoding="utf-8") as f:
    data = json.load(f)
print(type(data))  
print(data.keys())  
print(type(data["licitaciones"])) 
print(len(data["licitaciones"]))  
```

json.load() lee el fichero y devuelve directamente un diccionario Python. No devuelve una cadena: la conversacion JSON -> Phyton ya está hecha.

#### Qué es objeto y que es array

· El documento raíz es un objeto JSON ➡️se convierte en dict.

· meta es in objeto JSON ➡️ se convierte en dict

· licitaciones es un array JSON ➡️ se convuerte en list.

· Cada elemento de licitaciones es un objeto JSON ➡️ se convierte dict.

· Los valores como id, organo, objeto, moneda, cpv, fechaLimite y estado son cadenas ➡️ str

· importe, total y lotes son números ➡️ normalmente float o int.
